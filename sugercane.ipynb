```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df=pd.read_csv("/content/List of Countries by Sugarcane Production.csv")
```

```python
df.shape
```

```python
df.head()
```

```python
# data cleaning  trying to clean false statement in the sugercane production Tons
df['Production (Tons)']=df['Production (Tons)'].str.replace(".","")
```

```python
df.head()
```

```python
# data cleaning  trying to clean false statement in the sugercane production per person  entered  data wrong clean
df['Production per Person (Kg)']=df['Production per Person (Kg)'].str.replace(".","").str.replace(",","")
```

```python
df.head()
```

```python
# data cleaning  trying to clean false statement in the sugercane acreage
df['Acreage (Hectare)']=df['Acreage (Hectare)'].str.replace(".","")
```

```python
df.head()
```

```python
# data cleaning  trying to clean false statement in the sugercane yield
df['Yield (Kg / Hectare)']=df['Yield (Kg / Hectare)'].str.replace(".","")
```

```python
df.head()
```

```python
# data cleaning  trying to clean false statement in the sugercane yield
df['Yield (Kg / Hectare)']=df['Yield (Kg / Hectare)'].str.replace(",","")
```

```python
df.head()
```

```python
# Rename the column name with a good name to avoid naming issues
df.rename(columns={"Production (Tons)":"Production(Tons)"},inplace=True)
df.rename(columns={"Production per Person (Kg)":"ProductionperPerson(Kg)"},inplace=True)
df.rename(columns={"Acreage (Hectare)":"Acreage(Hectare)"},inplace=True)
df.rename(columns={"Yield (Kg / Hectare)":"Yield(Kg/Hectare)"},inplace=True)
```

```python
df.head()
```

```python
# Find the null raws
df.isnull().sum()
```

```python
df[df["Acreage(Hectare)"].isnull()]
```

```python
# in the data in acreage one contain null option
df=df.dropna()
```

```python
# drop unmamed index
df.drop([col for col in ["index", "Unnamed: 0"] if col in df.columns], axis=1, inplace=True)
```

```python
# Find the null raws
df.isnull().sum()
```

```python
df.head()
```

```python
df.dtypes
```

```python
df['Production(Tons)']=df['Production(Tons)'].astype(float)
df['ProductionperPerson(Kg)']=df['ProductionperPerson(Kg)'].astype(float)
df['Acreage(Hectare)']=df['Acreage(Hectare)'].astype(float)
df['Yield(Kg/Hectare)']=df['Yield(Kg/Hectare)'].astype(float)
```
